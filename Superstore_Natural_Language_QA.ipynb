## Predefined Questions for Q&A Testing

The following predefined business questions were created to test the natural-language Q&A engine. These questions cover key areas of the Superstore dataset, including sales, profit, categories, states, customers, and overall business metrics.

```python
def answer_question(question):

    question = question.lower().strip()

    # Highest sales sub-category
    if ("sub-category" in question or "sub category" in question) and "highest sales" in question:

        result = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False)

        sub_category = result.index[0]
        sales = result.iloc[0]

        return f"{sub_category} has the highest sales with approximately {sales:,.2f} USD."


    # Highest profit sub-category
    elif ("sub-category" in question or "sub category" in question) and "highest profit" in question:

        result = df.groupby("Sub-Category")["Profit"].sum().sort_values(ascending=False)

        sub_category = result.index[0]
        profit = result.iloc[0]

        return f"{sub_category} has the highest profit with approximately {profit:,.2f} USD."


    # Highest sales category
    elif "highest sales category" in question or "category has the highest sales" in question:

        result = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

        category = result.index[0]
        sales = result.iloc[0]

        return f"{category} has the highest sales with approximately {sales:,.2f} USD."


    # Highest profit category
    elif "highest profit category" in question or "category has the highest profit" in question:

        result = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

        category = result.index[0]
        profit = result.iloc[0]

        return f"{category} has the highest profit with approximately {profit:,.2f} USD."


    # Highest sales state
    elif "highest sales state" in question or "state has the highest sales" in question:

        result = df.groupby("State")["Sales"].sum().sort_values(ascending=False)

        state = result.index[0]
        sales = result.iloc[0]

        return f"{state} has the highest sales with approximately {sales:,.2f} USD."


    # Highest profit state
    elif "highest profit state" in question or "state has the highest profit" in question:

        result = df.groupby("State")["Profit"].sum().sort_values(ascending=False)

        state = result.index[0]
        profit = result.iloc[0]

        return f"{state} has the highest profit with approximately {profit:,.2f} USD."


    # Total sales
    elif "total sales" in question:

        total_sales = df["Sales"].sum()

        return f"Total sales are approximately {total_sales:,.2f} USD."


    # Total profit
    elif "total profit" in question:

        total_profit = df["Profit"].sum()

        return f"Total profit is approximately {total_profit:,.2f} USD."


    # Total quantity
    elif "total quantity" in question or "total units" in question:

        total_quantity = df["Quantity"].sum()

        return f"Total quantity sold is {total_quantity:,.0f} units."


    # Total orders
    elif "total orders" in question or "number of orders" in question:

        total_orders = df["Order ID"].nunique()

        return f"There are {total_orders:,} unique orders."


    # Total customers
    elif "total customers" in question or "number of customers" in question:

        total_customers = df["Customer ID"].nunique()

        return f"There are {total_customers:,} unique customers."


    # Average order value
    elif "average order value" in question or "aov" in question:

        total_sales = df["Sales"].sum()
        total_orders = df["Order ID"].nunique()

        aov = total_sales / total_orders

        return f"The average order value is approximately {aov:,.2f} USD."


    # Profit margin
    elif "profit margin" in question:

        total_sales = df["Sales"].sum()
        total_profit = df["Profit"].sum()

        margin = (total_profit / total_sales) * 100

        return f"The overall profit margin is approximately {margin:.2f}%."


    # Highest sales customer
    elif "highest sales customer" in question or "customer has the highest sales" in question:

        result = df.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False)

        customer = result.index[0]
        sales = result.iloc[0]

        return f"{customer} has the highest sales with approximately {sales:,.2f} USD."


    # Highest profit customer
    elif "highest profit customer" in question or "customer has the highest profit" in question:

        result = df.groupby("Customer Name")["Profit"].sum().sort_values(ascending=False)

        customer = result.index[0]
        profit = result.iloc[0]

        return f"{customer} has the highest profit with approximately {profit:,.2f} USD."


    # Biggest loss-making customer
    elif "highest loss customer" in question or "biggest loss customer" in question:

        result = df.groupby("Customer Name")["Profit"].sum().sort_values()

        customer = result.index[0]
        loss = result.iloc[0]

        return f"{customer} has the highest loss of approximately {abs(loss):,.2f} USD."


    # Biggest loss-making state
    elif "highest loss state" in question or "biggest loss state" in question:

        result = df.groupby("State")["Profit"].sum().sort_values()

        state = result.index[0]
        loss = result.iloc[0]

        return f"{state} has the highest loss of approximately {abs(loss):,.2f} USD."


    # Biggest loss-making sub-category
    elif ("sub-category" in question or "sub category" in question) and (
        "highest loss" in question or "biggest loss" in question
    ):

        result = df.groupby("Sub-Category")["Profit"].sum().sort_values()

        sub_category = result.index[0]
        loss = result.iloc[0]

        return f"{sub_category} has the highest loss of approximately {abs(loss):,.2f} USD."


    # Highest sales year
    elif "highest sales year" in question or "year has the highest sales" in question:

        result = df.groupby("Year")["Sales"].sum().sort_values(ascending=False)

        year = result.index[0]
        sales = result.iloc[0]

        return f"{year} has the highest sales with approximately {sales:,.2f} USD."


    # Highest profit year
    elif "highest profit year" in question or "year has the highest profit" in question:

        result = df.groupby("Year")["Profit"].sum().sort_values(ascending=False)

        year = result.index[0]
        profit = result.iloc[0]

        return f"{year} has the highest profit with approximately {profit:,.2f} USD."


    # Highest sales month
    elif "highest sales month" in question or "month has the highest sales" in question:

        result = df.groupby("Month_Name")["Sales"].sum().sort_values(ascending=False)

        month = result.index[0]
        sales = result.iloc[0]

        return f"{month} has the highest sales with approximately {sales:,.2f} USD."


    # Highest profit month
    elif "highest profit month" in question or "month has the highest profit" in question:

        result = df.groupby("Month_Name")["Profit"].sum().sort_values(ascending=False)

        month = result.index[0]
        profit = result.iloc[0]

        return f"{month} has the highest profit with approximately {profit:,.2f} USD."


    # Highest discount category
    elif "highest discount category" in question or "category has the highest discount" in question:

        result = df.groupby("Category")["Discount"].mean().sort_values(ascending=False)

        category = result.index[0]
        discount = result.iloc[0] * 100

        return f"{category} has the highest average discount at approximately {discount:.2f}%."


    # Average shipping days
    elif "average shipping days" in question or "average shipping time" in question:

        avg_shipping = df["Shipping Days"].mean()

        return f"The average shipping time is approximately {avg_shipping:.2f} days."


    else:
        return "Sorry, I don't understand this question yet."
```

```python
answer_question("Which sub-category has the highest sales?")
answer_question("Which sub-category has the highest profit?")
answer_question("Which category has the highest sales?")
answer_question("Which state has the highest profit?")
answer_question("What are the total sales?")
answer_question("What is the average order value?")
answer_question("Which customer has the highest profit?")
```

## Interactive Natural-Language Q&A

The Q&A engine allows users to enter business questions in natural language. The system processes the user's question, identifies the relevant analysis, and returns a data-driven answer from the Superstore dataset.

The user can continue asking questions until `exit` is entered.

```python
while True:

    question = input("\nYou: ")

    if question.lower() == "exit":
        print("Thank you for using the Superstore Data Analyst.")
        break

    answer = answer_question(question)

    print("Answer:", answer)
```
